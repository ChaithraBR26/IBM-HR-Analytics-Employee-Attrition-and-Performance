## Phase 1 — Data Understanding

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

### Display first and last records

```python
print('First 5 rows:')
display(df.head())
print('\nLast 5 rows:')
display(df.tail())
```

### Check rows and columns

```python
print(f'Number of rows and columns: {df.shape}')
print(f'Column names: {df.columns.tolist()}')
```

### Data types and Statistical summary

```python
print('Data types:')
df.info()
print('\nStatistical summary:')
display(df.describe())
```

## Phase 2 — Data Cleaning

### Check for Missing Values

```python
print('Missing values per column:')
df.isnull().sum()
```

### Check for Duplicate Rows

```python
print(f'Number of duplicate rows: {df.duplicated().sum()}')
if df.duplicated().sum() > 0:
    print('Dropping duplicate rows.')
    df.drop_duplicates(inplace=True)
    print(f'Number of rows after dropping duplicates: {df.shape[0]}')
```

## Phase 3 — Exploratory Data Analysis (Pandas)

### Employees by Department

```python
print('Employee count by Department:')
display(df['Department'].value_counts())
```

### Employees by Gender

```python
print('Employee count by Gender:')
display(df['Gender'].value_counts())
```

### Attrition Count

```python
print('Attrition count:')
display(df['Attrition'].value_counts())
```

### Average, Highest, and Lowest Monthly Income

```python
print(f'Average Monthly Income: {df["MonthlyIncome"].mean():.2f}')
print(f'Highest Monthly Income: {df["MonthlyIncome"].max()}')
print(f'Lowest Monthly Income: {df["MonthlyIncome"].min()}')
```

### Job Role Distribution

```python
print('Job Role Distribution:')
display(df['JobRole'].value_counts())
```

### Attrition by Department

```python
print('Attrition by Department:')
display(df.groupby('Department')['Attrition'].value_counts().unstack(fill_value=0))
```

### Attrition by Gender

```python
print('Attrition by Gender:')
display(df.groupby('Gender')['Attrition'].value_counts().unstack(fill_value=0))
```

## Phase 4 — NumPy

### Monthly Income Statistics using NumPy

```python
monthly_income = df['MonthlyIncome'].values

print(f'NumPy Mean Monthly Income: {np.mean(monthly_income):.2f}')
print(f'NumPy Median Monthly Income: {np.median(monthly_income):.2f}')
print(f'NumPy Standard Deviation of Monthly Income: {np.std(monthly_income):.2f}')
print(f'NumPy 25th Percentile of Monthly Income: {np.percentile(monthly_income, 25):.2f}')
print(f'NumPy 75th Percentile of Monthly Income: {np.percentile(monthly_income, 75):.2f}')
```

### Categorizing Monthly Income using `np.where`

```python
df['Income_Category'] = np.where(df['MonthlyIncome'] > 10000, 'High', 'Low')
print('Distribution of Income Categories:')
display(df['Income_Category'].value_counts())
```

## Phase 5 — Matplotlib

### Department Distribution (Bar Chart)

```python
plt.figure(figsize=(10, 6))
df['Department'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Department Distribution')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Monthly Income Distribution (Histogram)

```python
plt.figure(figsize=(10, 6))
df['MonthlyIncome'].hist(bins=30, color='lightcoral', edgecolor='black')
plt.title('Monthly Income Distribution')
plt.xlabel('Monthly Income')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Age Distribution (Histogram)

```python
plt.figure(figsize=(10, 6))
df['Age'].hist(bins=20, color='lightgreen', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Attrition Percentage (Pie Chart)

```python
plt.figure(figsize=(8, 8))
attrition_counts = df['Attrition'].value_counts()
plt.pie(attrition_counts, labels=attrition_counts.index, autopct='%1.1f%%', startangle=90, colors=['#66b3ff','#99ff99'])
plt.title('Attrition Percentage')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()
```

## Phase 6 — Seaborn

### Salary vs Attrition (Boxplot)

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df, palette='pastel')
plt.title('Monthly Income vs. Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Age vs Attrition (Violinplot)

```python
plt.figure(figsize=(8, 6))
sns.violinplot(x='Attrition', y='Age', data=df, palette='viridis')
plt.title('Age vs. Attrition')
plt.xlabel('Attrition')
plt.ylabel('Age')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Department-wise Attrition (Countplot)

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Department', hue='Attrition', data=df, palette='coolwarm')
plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Correlation Heatmap

```python
plt.figure(figsize=(12, 10))
# Select only numerical columns for correlation
numerical_df = df.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numerical_df.corr(), annot=False, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```
